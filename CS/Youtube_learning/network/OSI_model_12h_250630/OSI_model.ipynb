How email works

SMTP(SMTP stands for Simple Mail Transfer Protocol. Its one and only job is to send or push emails), PoP3(Post Office Protocol version 3. Its job is to retrieve or pull emails from a mail server.)

You Hit "Send": You compose a message in your email client (like Gmail or Outlook) and hit send. Your email client is like your local post office.

Sending the Mail (SMTP): Your email client uses SMTP to send your message to your outgoing mail server. This is like the mail truck picking up your letter from the post office.

Routing the Mail (SMTP): Your mail server then acts as a sorting facility. It looks at the recipient's address (e.g., jane@example.com), finds the mail server for "example.com," and uses SMTP again to send your email across the internet to that specific server.

Mail Arrives at the Destination Server: The recipient's mail server receives the email and places it in their mailbox, where it waits to be picked up.

Picking Up the Mail (POP3): When the recipient opens their email client, the client uses POP3 to connect to their mail server and download your message to their device. This is like them going to their mailbox and taking the letter out.

a DNS (Domain Name System) lookup
The User's Request: You type google.com into your browser and hit enter. Your computer sends a request to its designated DNS resolver, which is usually managed by your ISP (Internet Service Provider). The question is, "What is the IP address for google.com?"

ISP to Root Server: Your ISP's resolver doesn't know the address, so it starts by asking one of the internet's Root Servers. It asks, "Where can I find information about the .com domain?"

Root Server's Reply: The Root Server doesn't know the IP address for google.com, but it knows who is in charge of all .com addresses. It replies, "I don't know, but you should ask the TLD (Top-Level Domain) server that manages .com."

ISP to TLD Server: Your ISP's resolver now asks the .com TLD server: "What is the IP address for google.com?"

TLD Server's Reply: The TLD server knows which specific servers hold the official records for google.com (these are called authoritative nameservers). It looks up the record and sends the final IP address for google.com back to your ISP's resolver.

ISP to Your Computer: Your ISP's resolver now has the correct IP address. It sends this address back to your computer.

transport layer
Preventing Network Overload (Congestion Control)
Managing Multiple Applications (Multiplexing)
Multiplexing: On the sending side, the Transport Layer takes data streams from different applications (a message, a file, a video conference) and combines them into a single flow of packets to be sent over the network.

Demultiplexing: On the receiving side, it separates these packets and, by reading the port numbers it attached, delivers the data to the correct application. This ensures that the message goes to your chat app, the file goes to your downloads folder, and the video goes to your video conference app.

checksum( error detection):
    Sender: Before sending a packet, the sender calculates the checksum and attaches it to the packet's header.
    Receiver: When the packet arrives, the receiver performs the exact same calculation on the received data.

timers(timer start, timer end, and timer expire): handling lost packets
    Timer Start: The "stopwatch" begins as soon as the packet is sent.
    Timer End: The timer stops when the sender receives a successful acknowledgment from the receiver.
    Timer Expire: If the timer runs out before an acknowledgment arrives, the sender assumes the packet was lost. It then retransmits the same packet and starts a new timer.

sequence numbers:
    A sequence number is a unique number assigned to each data packet. It serves two critical purposes:
    
    Reordering Packets: Data packets can travel across the internet and arrive out of order. The receiver uses the sequence numbers to reassemble the packets into their original, correct sequence before passing the data to the application.Tracking and Acknowledgment: The receiver uses sequence numbers to tell the sender which packets it has successfully received. This allows the sender to know which packets, if any, were lost and need to be resent.

TCP (Transmission Control Protocol)

Breaking Down and Rebuilding Data (Segmentation): It divides the data into chunks: TCP's first job is to take this large data stream and break it down into smaller, numbered pieces called segments.

It adds headers: Each segment gets a header attached, which contains crucial information like the source and destination port numbers, sequence numbers, and a checksum for error checking. This is like putting letters into separate, addressed envelopes.

Ensuring Reliable Delivery
Congestion Control: TCP actively monitors the network for "traffic jams." If it detects congestion, it slows down the data transmission rate to prevent the network from being overwhelmed and dropping packets.

Takes care of arrival: TCP makes sure the data arrives successfully. It tracks each segment and waits for an acknowledgment from the receiver. If a segment "does not arrive" (gets lost), TCP will retransmit it to ensure the recipient gets the complete data.

one TCP connection is only for 2 computers

Full-duplex: means that once a connection is established, data can be sent and received by both sides at the same time over the same connection.

TCP 3-Way Handshake 

Step 1: SYN (Synchronize)
The Client initiates the connection by sending a SYN packet to the Server. This packet includes a starting Sequence Number (in the diagram, Seq no: 32), which is a random number used to track the data it sends.

Step 2: SYN-ACK (Synchronize-Acknowledge)
The Server responds with a single SYN-ACK packet that does two things:

SYN: It sends its own starting Sequence Number (in the diagram, ex: 56).

ACK: It acknowledges the client's message by sending an Acknowledgment Number that is the client's sequence number plus one (Ack no: 33).

Step 3: ACK (Acknowledge)
The Client sends a final ACK packet to the Server. It acknowledges the server's message by setting its Acknowledgment Number to the server's sequence number plus one (Ack no: 57).

router: handles logical device addressing and determines the best path for data to travel from its source to its destination across multiple networks. Its primary protocol is the Internet Protocol (IP).

routing table: A routing table is a set of rules, much like a map or a GPS database, stored inside a router. It lists the known network destinations and specifies the best path to forward packets to reach those destinations.

This is different from the data plane, which is the "muscle" that actually forwards the packets based on the decisions made by the control plane.

Internet Protocols

IPv4: 32bits, 4 words
ipv6: 128 bits, 16 words

IPv4: The older, traditional version. It uses 32-bit addresses, which provides about 4.3 billion unique addresses. They are written as four numbers separated by dots (e.g., 172.217.14.238). The "4 words" in the note refers to these four numbers.

IPv6: The newer version created to solve the shortage of IPv4 addresses. It uses 128-bit addresses, providing an almost unimaginably large number of unique addresses for the future of the internet.

A loopback address (any address beginning with 127, like 127.0.0.1) is a special IP address that always points back to your own computer.

It's used by developers and network engineers to test applications on a machine without sending any traffic out onto the actual network. The universally recognized hostname for this address is localhost.

"TTL" (Time To Live)
which is a crucial mechanism for preventing data packets from circulating endlessly on the internet in a routing loop.

Here's how it works:

A packet is sent with a starting TTL value (e.g., 64).

Each router that forwards the packet subtracts at least one from the TTL counter.

If the TTL counter reaches zero, the router discards the packet.

network middlebox
Firewall
Its primary job is to filter IP packets based on a set of security rules. The firewall inspects passing traffic and decides whether to block it or allow it based on criteria such as:

IP Addresses: Blocking traffic from known malicious addresses.
Port Numbers: Allowing traffic only for specific services (e.g., allowing web traffic on port 443 but blocking other ports).
Protocols: Permitting certain protocols (like TCP) while blocking others.
Flags: Analyzing TCP flags to check the state of a connection.

Network Address Translation (NAT): a common firewall function where it changes the private IP address of a device on your local network to a public IP address before sending the packet to the internet.

NAT is a method used by a router or firewall to allow multiple devices in a private network (using private IP addresses like 192.168.x.x) to share a single public IP address to access the internet. (change source ID or destination ID in the header)
This has two main benefits:
Conserves public IP addresses.

Hides the internal network structure, providing a basic layer of security.

Stateless Firewall 
A stateless firewall inspects each data packet in isolation and makes a block or allow decision based on a fixed set of rules. It has no memory of past connections.

Stateful Firewall 
A stateful firewall is a more intelligent firewall that monitors the state of active connections. It keeps a "state table" that remembers the details of every outgoing connection.

Data link layer

data link layer address(MAC address): many devices connected in LAN (local area network)

Address Resolution Protocol (ARP).

On a network, devices have two main types of addresses:
An IP address (Layer 3), which is a logical address that can change.
A MAC address (Data Link Layer address), which is a permanent, physical hardware address.


address resolution: the process of finding a device's MAC address when you only know its IP address.
The process works like this:

A device (e.g., Computer 1) wants to send data to another device (e.g., Computer 3) and knows its IP address.
Computer 1 broadcasts an ARP request to every device on the LAN, asking, "Who has this IP address?"
Computer 3 recognizes its IP address and sends a reply directly to Computer 1, saying, "I have that IP, and here is my MAC address."

All devices on the LAN process the ARP Request, but only the host whose IP address matches the target IP in the request will generate a response. This target host creates an ARP Reply packet containing its own MAC address.

### The Complete Story: From Your Computer to the Web

Imagine you open your laptop to browse the web. Here’s the entire journey, connecting all the concepts.

---

### ## 1. Getting Online: The Connection

First, your computer needs to get on the internet.

* **DHCP**: Your computer connects to your Wi-Fi router. The first thing it does is ask for an IP address. **DHCP** is the protocol that automatically assigns your computer a unique **private IP address** (like `192.168.1.102`), allowing it to talk to other devices on your home network.
* **ISP**: Your connection to the wider internet is provided by your **ISP** (Internet Service Provider) like Comcast or AT&T. They manage the physical infrastructure (like **SONET** fiber optic cables) that forms the on-ramp to the global network.
* **NAT & Firewall**: Your home router performs **NAT** (Network Address Translation), taking all the devices in your home and letting them share a single **public IP address** provided by your ISP. Your router also acts as a **stateful firewall**, a security guard that blocks unsolicited incoming traffic while remembering and allowing the traffic you request.

---

### ## 2. Finding a Website: The DNS Lookup

You type `google.com` into your browser. Your computer doesn't know where that is, so it needs to find the server's IP address.

* **DNS**: Your computer sends a **DNS** (Domain Name System) request to your ISP. This process is like looking up a name in the internet's phonebook.
* The request goes from your ISP's resolver to **Root Servers**, then **TLD Servers** (for `.com`), and finally to Google's authoritative nameserver, which replies with the correct public IP address for `google.com`.

---

### ## 3. Making the Connection: The Handshake

Now that your computer has the IP address, it needs to establish a reliable connection with Google's server.

* **TCP**: It uses **TCP** (Transmission Control Protocol), the reliable "registered mail" service of the internet.
* **3-Way Handshake**: A three-step negotiation (`SYN`, `SYN-ACK`, `ACK`) occurs to ensure both sides are ready.
* **Sockets and Ports**: A connection **socket** is created, which is the combination of Google's IP address and a specific **port** (like port 443 for secure web traffic). This TCP connection is **full-duplex**, meaning data can be sent and received at the same time.

---

### ## 4. Sending the Data: The Journey of a Packet

Your browser is now ready to ask for the webpage.

* **HTTP**: It creates an **HTTP** `GET` request.
* **TCP Segmentation**: TCP breaks this request into smaller, numbered pieces called **segments**. It adds a TCP header containing **sequence numbers** (to reorder them correctly on arrival) and a **checksum** (for error checking). This extra information is called **overhead**.
* **IP Packets**: Each segment is placed inside an **IP packet**. An IP header is added, containing the source and destination IP addresses and a **TTL** (Time To Live) to prevent the packet from looping forever.
* **ARP & Frames**: To send the packet to your local router, your computer uses **ARP** (Address Resolution Protocol) to find the router's physical **MAC address**. The IP packet is then wrapped in an Ethernet **frame**, which is addressed with the MAC address for the next hop.

---

### ## 5. Navigating the Internet: Routing

The packets travel from your home router across the internet.

* **Routers and Routing Tables**: Each **router** along the path reads the destination IP address on the packet. It consults its **routing table** (built by **dynamic routing** protocols) to decide the next best hop to forward the packet to.
* This process repeats across multiple routers until the packets reach their destination at Google's data center. The firewall at Google inspects the incoming packets before allowing them in.

---

### ## 6. The Response and Reliability

Google's server receives your request, assembles the packets, and sends the website's data back to you using the same process.

* **Reliability**: If any TCP packets get lost, the **timers** on the sender's end will expire, and the lost packet will be retransmitted. If the network becomes too busy, TCP's **congestion control** will slow down the transfer to prevent a traffic jam.
* **UDP**: For services where speed is more important than perfect reliability (like live video streaming or online games), the internet uses **UDP** (User Datagram Protocol), the fast "postcard" service.

---

### ## 7. The Web Experience: Cookies

Your browser receives all the data packets, reassembles them, and displays the Google homepage.

* **Cookies**: If you log into your Google account, the server sends back a **cookie**, a small text file that your browser stores. This cookie helps the server remember who you are so you don't have to log in on every single page. This is a **first-party cookie**. Any tracking cookies from ad networks on the page would be **third-party cookies**.
