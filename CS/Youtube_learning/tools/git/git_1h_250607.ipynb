git status
git status -s // show the staging area

git add *.txt
git add file2.txt file1.txt
git rm file2.txt file1.txt // remove files from both working directory and staging area

mkdir logs // create a new directory called logs
echo hello > logs/dev.log // write a line to somewhere
echo logs/ > .gitignore // move it to gitignore(pattern *.log)

git ls-files // show all the files in staging area

git rm --cached -r bin/  // remove entire something from staging area

git diff --staged // see the differences of the files between last commit and current staging area

git diff // see unstaged changes

git log --oneline (--reverse) // show all commits with oneline description, from latest to oldest

git show (commit id)/(HEAD~number)// show one full commit

git ls-tree HEAD~1 // show blob(file), tree(directory)

git restore —staged (files) // restore the files in staging area

git restore . // restore all the files (undo all local changes)

git clean -fd // force to remove whole directory
