```python
git status # Shows the current state of your working directory and staging area. 
git status -s # git status --short: Gives a more compact, short-format output. Useful for quickly seeing which files have changed.
git status -v # git status --verbose: Shows more detailed information, including the textual changes of uncommitted files

git diff  #Shows the differences between various states in your Git repository. 
git diff <commit1_hash> <commit2_hash> #shows the changes between two specific commits.
git diff <branch1_name> <branch2_name> #shows the differences between the tips of two branches.
# A specific file: You can append a file path to any of the above commands to see differences only for that file

git add # add changes to staging area
```

```python
git push origin main

git pull

git fetch # Downloads changes from a remote repository to your local repository but does not automatically merge them into your current working branch.
# These updates are stored in your local repository as remote-tracking branches (e.g., origin/main, origin/feature-branch). Your local branches (e.g., main, feature-branch) remain untouched.
# You can checkout a remote-tracking branch (e.g., git checkout origin/main) to inspect the state of the remote branch without affecting your local branches. 

git stash # Temporarily saves your uncommitted local changes (both staged and unstaged) so you can switch context (e.g., switch branches, pull updates, work on an urgent fix) without committing incomplete work.
git stash list #Shows all the stashes you've created. They are typically named stash@{0}, stash@{1}, etc., with stash@{0} being the most recent.
git stash apply [<stash>] #Reapplies the changes from a specific stash (or the latest one if unspecified) to your working directory. The stash itself is kept in the stash list.
git stash pop [<stash>] # Similar to apply, but it also removes the stash from the list after successfully applying it.
git stash drop [<stash>] # Deletes a specific stash from the list.
git stash clear # Deletes all stashes.
```

Workflows

Feature Branching
Core Idea: All development for a new feature happens in a dedicated branch created specifically for that feature, rather than directly on the main branch (often called main or master).
![pic1](images/feature_branch.png)

Gitflow
Core Idea: A more structured and stricter branching model than simple Feature Branching, designed around project releases. It introduces several types of branches with specific roles and interactions. It's often considered a more "legacy" workflow, especially for web apps with continuous delivery, but can still be useful for projects with scheduled releases and multiple versions to support.
![pic2](images/gitflow.png)

GitHub Flow
Core Idea: A lightweight, branch-based workflow that emphasizes simplicity and continuous deployment. It's simpler than GitFlow and is what GitHub itself uses.
![pic3](images/githubflow.png)

GitLab Flow
Core Idea: A workflow that aims to be simpler than GitFlow but more structured than GitHub Flow, particularly by incorporating environment-specific or release-specific branches. It strongly emphasizes integrating Git with issue tracking.
![pic4](images/gitlabflow.png)

Trunk-Based Dev
Core Idea: A source-control branching model where developers collaborate on code in a single branch called "trunk" (commonly main or master in Git). The emphasis is on avoiding long-lived development branches and integrating changes frequently, often multiple times a day per developer.
![pic5](images/trunk.png)
