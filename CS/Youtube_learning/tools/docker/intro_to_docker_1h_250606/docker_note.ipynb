docker images
docker ps // see the running containers
docker ps -a // see all the containers


docker pull [name]:[tag] // if tags, latest version // can use run directly
docker run [name]:[tag]
docker run -d [name]:[tag] // detach, it will run without blocking the terminal

docker logs [container id] // in detach mode, can see logs

docker stop [container id]/[container_name] // stop running containers

docker run -d -p [host_port]:[container_port] [name]:[tag] // binding port to host_port, only one service can run on one specific localhost port
// standard: same port number with container port

docker start [container_id] // start stopped container


docker run --name [name of container]  -d -p [host_port]:[container_port] [name]:[tag] // create a container with a specific name

docker build -t [name]:[tag] [location of dockerfile]
